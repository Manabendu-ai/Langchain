## Tools

#### models can request to call tools that perform tasks such as fetching data from a database, searching the web, or running code.
#### Tools are a pair of:
* #### Schema : includes the name of the tool, a description, and/or argurment definition(Often a JSON)
* #### A function or Coroutine to execute.

```python
import os
from langchain_groq import ChatGroq

model = ChatGroq(
     model = "llama-3.3-70b-versatile"
)

response = model.invoke("what type of engine is used in Lamborghini Urus?")
response.content
```

```python
from langchain.tools import tool

@tool
def get_price(car:str ,model : str) -> str:
    """Get the Price of the specific Car Model"""
    return f"Price of {car}, {model} is {9999999}INR."

model_with_tools = model.bind_tools([get_price])
```

```python
response = model_with_tools.invoke("WHat is the price of Lamborghini Hurracine-V6?")
print(response)
for tool_call in response.tool_calls:
    print(f"Tool: {tool_call['name']}")
    print(f"Args: {tool_call['args']}")
```
