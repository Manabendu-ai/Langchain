### Streaming and Batch

#### Streaming

`Most models can stream their output content while being generated. By displaying output progressively,
streaming significantly improves user experience, particularly for longer responses. Calling stream() returns an iterator that yields
output chunks as they are produced. You can use a Loop to process each chunk in real-time:`

```python
import os
from langchain_groq import ChatGroq

os.environ["groq_api_key"] = os.getenv("GROQ_API_KEY")
model = ChatGroq(
    model = "llama-3.3-70b-versatile"
)
```

```python
for chunks in model.stream("Why do parrot's have colorful feathers?"):
    print(chunks.text, end="|", flush=True)
```

### Batch

`Batching is the collection of independent requests to a model that can significantly improve performance and reduce cost,
 as the processing can be done in parallel.`
